```python
def col_num_not_na(df: pd.DataFrame, col: str):
    return len(df[~df[col].isna()])
def col_num_unique_values(df: pd.DataFrame, col: str):
    return len(df[col].unique())
```

```python
print(f'len(df): {len(df)}')
for col in df.columns:
    print(f'---{col}---\n\t Number of entries with "{col}": {col_num_not_na(df, col)}\n\t Number of unique values: {col_num_unique_values(df, col)}')
```
