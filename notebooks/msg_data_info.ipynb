## Overlap: MassSpecGym vs ATMOMACCS dataset

```python
atmo_df = pd.read_csv('../data/atmomaccs/labels.tsv', sep='\t')

# MSG inchikeys are the 14-char connectivity block; normalise both to that
msg_keys  = set(labels_df['inchikey'].str[:14])
atmo_keys = set(atmo_df['inchikey'].str[:14])

overlap      = msg_keys & atmo_keys
only_msg     = msg_keys - atmo_keys
only_atmo    = atmo_keys - msg_keys

print(f"Unique MSG molecules:           {len(msg_keys):>6}")
print(f"Unique ATMOMACCS molecules:     {len(atmo_keys):>6}")
print(f"Overlap (in both):              {len(overlap):>6}  ({100*len(overlap)/len(atmo_keys):.1f}% of ATMOMACCS)")
print(f"Only in MSG:                    {len(only_msg):>6}")
print(f"Only in ATMOMACCS:              {len(only_atmo):>6}")
```

## ATMOMACCS fingerprinting + clustering of MSG molecules

```python
import sys
sys.path.insert(0, '../src')
from ATMOMACCS_no_binary import smiles_to_atmomaccs

fps = smiles_to_atmomaccs(labels_df['smiles2d'], as_dataframe=False)
print(f"Fingerprint matrix: {fps.shape}")  # (28924, fp_len)
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

N_CLUSTERS = 20

scaler = StandardScaler()
fps_scaled = scaler.fit_transform(fps)

kmeans = KMeans(n_clusters=N_CLUSTERS, random_state=42, n_init='auto')
labels_df = labels_df.copy()
labels_df['cluster'] = kmeans.fit_predict(fps_scaled)

print(labels_df['cluster'].value_counts().sort_index())
```

```python
import umap
import matplotlib.pyplot as plt

reducer = umap.UMAP(n_components=2, random_state=42)
embedding = reducer.fit_transform(fps_scaled)

# mark which MSG molecules are also in the ATMOMACCS dataset
in_atmo = labels_df['inchikey'].str[:14].isin(overlap)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# left: colour by cluster
sc = axes[0].scatter(embedding[:, 0], embedding[:, 1],
                     c=labels_df['cluster'], cmap='tab20',
                     s=2, alpha=0.5)
axes[0].set_title('MSG molecules — ATMOMACCS clusters')
plt.colorbar(sc, ax=axes[0], label='cluster')

# right: highlight overlap with ATMOMACCS dataset
axes[1].scatter(embedding[~in_atmo, 0], embedding[~in_atmo, 1],
                c='steelblue', s=2, alpha=0.3, label='MSG only')
axes[1].scatter(embedding[in_atmo, 0], embedding[in_atmo, 1],
                c='crimson', s=8, alpha=0.9, label=f'In ATMOMACCS ({in_atmo.sum()})')
axes[1].set_title('Overlap with ATMOMACCS dataset')
axes[1].legend(markerscale=4)

plt.tight_layout()
plt.show()
```

```python
import sys
import pandas as pd
import numpy as np
from rdkit import Chem

sys.path.insert(0, '../src')
from ATMOMACCS_no_binary import smiles_to_atmomaccs

# --- load & deduplicate ---
msg_df = labels_df[['smiles2d', 'inchikey']].copy()
msg_df['dataset'] = 'MSG'

atmo_df = pd.read_csv('../data/atmomaccs/labels.tsv', sep='\t')
atmo_df['smiles2d'] = atmo_df['smiles'].apply(
    lambda s: Chem.MolToSmiles(Chem.MolFromSmiles(s), isomericSmiles=False)
)
#atmo_df = atmo_df[~atmo_df['spec'].str.contains('ferraz-caetano')]
atmo_df = atmo_df.rename(columns={'inchikey': 'inchikey'})[['smiles2d', 'inchikey', 'dataset']]
combined = pd.concat([msg_df, atmo_df], ignore_index=True)
combined = combined.drop_duplicates(subset='smiles2d').reset_index(drop=True)
print(f"Combined unique molecules: {len(combined)}  "
      f"(MSG: {(combined['dataset']=='MSG').sum()}, "
      f"ATMOMACCS: {(combined['dataset']=='atmomaccs').sum()})")

# --- fingerprints ---
fps = smiles_to_atmomaccs(combined['smiles2d'])
print(f"Fingerprint matrix: {fps.shape}")

# --- cluster ---
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

N_CLUSTERS = 30
fps_scaled = StandardScaler().fit_transform(fps)
combined['cluster'] = KMeans(n_clusters=N_CLUSTERS, random_state=42, n_init='auto').fit_predict(fps_scaled)

# --- UMAP + plot ---
import umap
import matplotlib.pyplot as plt

embedding = umap.UMAP(n_components=2, random_state=42).fit_transform(fps_scaled)
```

```python
colors = {'MSG': 'steelblue', 'atmomaccs': 'crimson'}
fig, ax = plt.subplots(figsize=(10, 7))
for ds, grp in combined.groupby('dataset'):
    idx = grp.index
    ax.scatter(embedding[idx, 0], embedding[idx, 1],
               c=colors[ds], label=ds, s=25)

ax.set_title('ATMOMACCS fingerprint UMAP: MSG vs ATMOMACCS dataset')
ax.legend(markerscale=2)
plt.tight_layout()
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

N_CLUSTERS = 20

fps_scaled = StandardScaler().fit_transform(fps)
combined['cluster'] = KMeans(n_clusters=N_CLUSTERS, random_state=42, n_init='auto').fit_predict(fps_scaled)

# fraction of each cluster that is ATMOMACCS
cluster_stats = combined.groupby('cluster')['dataset'].value_counts().unstack(fill_value=0)
cluster_stats['atmo_fraction'] = cluster_stats['atmomaccs'] / cluster_stats.sum(axis=1)
cluster_stats['total'] = cluster_stats.sum(axis=1)
print(cluster_stats.sort_values('atmo_fraction', ascending=False))

# bar chart
cluster_stats['atmo_fraction'].sort_values(ascending=False).plot(
    kind='bar', figsize=(10, 4),
    ylabel='Fraction ATMOMACCS', xlabel='Cluster',
    title='ATMOMACCS representation per cluster'
)
plt.tight_layout()
plt.show()
```

```python
# merge atmo_fraction back onto combined
combined['atmo_fraction'] = combined['cluster'].map(cluster_stats['atmo_fraction'])

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# left: colour by atmo_fraction of cluster
sc = axes[0].scatter(embedding[:, 0], embedding[:, 1],
                     c=combined['atmo_fraction'], cmap='RdYlBu_r',
                     s=3, alpha=0.5)
plt.colorbar(sc, ax=axes[0], label='ATMOMACCS fraction in cluster')
axes[0].set_title('Clusters coloured by ATMOMACCS fraction')

# right: colour by dataset, on top of cluster boundaries
is_atmo = combined['dataset'] == 'atmomaccs'
axes[1].scatter(embedding[~is_atmo, 0], embedding[~is_atmo, 1],
                c='steelblue', s=2, alpha=0.3, label='MSG')
axes[1].scatter(embedding[is_atmo, 0], embedding[is_atmo, 1],
                c='crimson', s=8, alpha=0.8, label='ATMOMACCS')
axes[1].set_title('Dataset origin')
axes[1].legend(markerscale=4)

plt.tight_layout()
plt.show()
```

```python
from rdkit import Chem
import pandas as pd
import numpy as np
import sys
sys.path.insert(0, '../src')
from ATMOMACCS_no_binary import smiles_to_atmomaccs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import umap
import matplotlib.pyplot as plt

# --- load & deduplicate ---
neims_df = pd.read_csv('../data/neims/labels.tsv', sep='\t')
neims_df['smiles2d'] = neims_df['smiles'].apply(
    lambda s: Chem.MolToSmiles(Chem.MolFromSmiles(s), isomericSmiles=False)
    if Chem.MolFromSmiles(s) else None
)
neims_df = neims_df.dropna(subset=['smiles2d'])
neims_df = neims_df.drop_duplicates(subset='smiles2d').reset_index(drop=True)
neims_df['dataset'] = 'NEIMS'

atmo_df = pd.read_csv('../data/atmomaccs/labels.tsv', sep='\t')
atmo_df['smiles2d'] = atmo_df['smiles'].apply(
    lambda s: Chem.MolToSmiles(Chem.MolFromSmiles(s), isomericSmiles=False)
    if Chem.MolFromSmiles(s) else None
)
atmo_df = atmo_df.dropna(subset=['smiles2d'])
atmo_df = atmo_df.drop_duplicates(subset='smiles2d').reset_index(drop=True)

combined = pd.concat(
    [neims_df[['smiles2d', 'inchikey', 'dataset']],
     atmo_df[['smiles2d', 'inchikey', 'dataset']]],
    ignore_index=True
).drop_duplicates(subset='smiles2d').reset_index(drop=True)

print(f"Combined: {len(combined)}  "
      f"(NEIMS: {(combined['dataset']=='NEIMS').sum()}, "
      f"ATMOMACCS: {(combined['dataset']=='atmomaccs').sum()})")

# --- overlap ---
neims_keys = set(neims_df['inchikey'].str[:14])
atmo_keys  = set(atmo_df['inchikey'].str[:14])
overlap    = neims_keys & atmo_keys
print(f"\nUnique NEIMS molecules:         {len(neims_keys):>6}")
print(f"Unique ATMOMACCS molecules:     {len(atmo_keys):>6}")
print(f"Overlap:                        {len(overlap):>6}  ({100*len(overlap)/len(atmo_keys):.1f}% of ATMOMACCS)")
print(f"Only in NEIMS:                  {len(neims_keys - atmo_keys):>6}")
print(f"Only in ATMOMACCS:              {len(atmo_keys - neims_keys):>6}")

# --- fingerprints + cluster ---
fps = smiles_to_atmomaccs(combined['smiles2d'])
fps_scaled = StandardScaler().fit_transform(fps)

N_CLUSTERS = 20
combined['cluster'] = KMeans(n_clusters=N_CLUSTERS, random_state=42, n_init='auto').fit_predict(fps_scaled)

cluster_stats = combined.groupby('cluster')['dataset'].value_counts().unstack(fill_value=0)
cluster_stats['atmo_fraction'] = cluster_stats['atmomaccs'] / cluster_stats.sum(axis=1)
combined['atmo_fraction'] = combined['cluster'].map(cluster_stats['atmo_fraction'])

# --- UMAP ---
from sklearn.decomposition import PCA

fps_pca = PCA(n_components=50).fit_transform(fps_scaled)
embedding = umap.UMAP(n_components=2, random_state=42).fit_transform(fps_pca)

is_atmo = combined['dataset'] == 'atmomaccs'
fig, axes = plt.subplots(1, 3, figsize=(22, 6))

sc = axes[0].scatter(embedding[:, 0], embedding[:, 1],
                     c=combined['atmo_fraction'], cmap='RdYlBu_r', s=3, alpha=0.5)
plt.colorbar(sc, ax=axes[0], label='ATMOMACCS fraction in cluster')
axes[0].set_title('Clusters coloured by ATMOMACCS fraction')

axes[1].scatter(embedding[~is_atmo, 0], embedding[~is_atmo, 1],
                c='steelblue', s=2, alpha=0.3, label='NEIMS')
axes[1].scatter(embedding[is_atmo, 0], embedding[is_atmo, 1],
                c='crimson', s=8, alpha=0.8, label='ATMOMACCS')
axes[1].set_title('NEIMS vs ATMOMACCS fingerprint UMAP')
axes[1].legend(markerscale=4)

sc = axes[2].scatter(embedding[:, 0], embedding[:, 1],
                     c=combined['atmo_fraction'], cmap='RdYlBu_r', s=3, alpha=0.3)
axes[2].scatter(embedding[is_atmo, 0], embedding[is_atmo, 1],
                c='crimson', s=8, alpha=0.9, label='ATMOMACCS')
plt.colorbar(sc, ax=axes[2], label='ATMOMACCS fraction in cluster')
axes[2].set_title('Fraction + ATMOMACCS points overlaid')
axes[2].legend(markerscale=4)

plt.tight_layout()
plt.show()
```
